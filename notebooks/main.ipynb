```python
# main.py
import sys
from pathlib import Path

import numpy as np

sys.path.insert(0, str(Path.cwd().parent / "src"))

from experiments import run_experiments, summarize, plot_error_boxplots, plot_position_estimates, plot_q_posteriors

scenarios = {
    "base":           dict(),
    "vento_forte":    dict(u=10.0),
    "fonte_distante": dict(source_position=(30.0, 25.0, 1.5)),
    "q_alto":         dict(q=5.0),
}

results = run_experiments(scenarios, seeds=range(50), keep_particles=True)   # teste rápido
summarize(results)
plot_error_boxplots(results)
plot_position_estimates(results, scenarios)
plot_q_posteriors(results, scenarios)

from DispersionModels import gaussian_plume_gp
from Sensors import PIDGasSensor
from MovementStrategies import ParameterizedSweepStrategy
from SimulationClasses import World, UAV, DispersionModelConfig, MultiUAVHistory
from Visualization import plot_mission_results, plot_particle_filter_step, plot_q_estimate
from ParticleFilter import ParticleFilter, ParticleFilterConfig
from Likelihood import gaussian_likelihood
from Resample import systematic_resample
import matplotlib.pyplot as plt

#Parâmetros verdadeiros: vento, fonte em (10, 10, 1.5), q=1, u=5, zetas=(0.3, 0.3).
def main():
    # --- 1. Global Parameters & World Setup ---
    true_pitch = 0.0
    true_yaw = np.arctan2(1.0, 1.0) # Retorna o ângulo correto no plano XY
    wind_angles_gt = (true_pitch, true_yaw)
    source_position=(10.0, 10.0, 1.5)

    # Nova estrutura de configuração do ambiente (Ground Truth)
    env_config = DispersionModelConfig(
        source_position=source_position,
        q=1.0,
        u=5.0,
        zetas=(0.3, 0.3),
        wind_angles=wind_angles_gt
    )

#Cria o World com o modelo e a config verdadeira.
    # O World recebe a função pura e o struct de configuração
    world = World(
        dispersion_model=gaussian_plume_gp, 
        model_config=env_config
    )

#Cria o UAV com varredura e dois sensores (MQ4 e CO2). Posição inicial em t=0.
    # --- 2. UAV & Sensors Setup ---
    strategy = ParameterizedSweepStrategy(x_max=40.0, y_max=40.0, num_sectors=12)
    uav1 = UAV("UAV_01", world, strategy)

    # Adding multiple named sensors to the UAV
    uav1.add_sensor(PIDGasSensor(name="MQ4", min_resolution=1e-9, noise_std=0.001))
    uav1.add_sensor(PIDGasSensor(name="CO2_Sensor", min_resolution=1e-3, noise_std=0.005))

    # Initialize UAV position
    t_progression = 0.0
    t_step = 0.005
    world.register_uav(uav1.id, strategy.compute_next_step(t_progression))


#Configura o PF (3000 partículas, limiar 0.5, sensor MQ4) e define as prioris. Inicializa as partículas.
    # --- 3. Estimator Setup ---
    pf_estimator = ParticleFilter()
    pf_config = ParticleFilterConfig(
        num_particles=3000,
        n_eff_threshold=0.5,
        resample_function=systematic_resample,
        likelihood_function=gaussian_likelihood,
        target_sensor_name="MQ4",
        dispersion_model=gaussian_plume_gp
    )

    # Definindo as distribuições a priori de forma hierarquizada
    # O Parametro dist não faz nada por hora...
    world_prior_params = {
        "x_s": {"dist": "uniform", "low": 0.0, "high": 40.0},
        "y_s": {"dist": "uniform", "low": 0.0, "high": 40.0},
        "q":   {"dist": "gamma", "shape": 1.5, "scale": 1.0},
        "u_s": {"dist": "normal", "mean": 1.0, "std": 0.2}, 
        "pitch": {"low": -np.pi/2, "high": np.pi/2}, # Pitch geralmente é restrito a [-90, 90] graus
        "yaw":   {"low": -np.pi, "high": np.pi},
        "zeta1": {"dist": "uniform", "low": 1.0, "high": 25.0},
        "zeta2": {"dist": "uniform", "low": 1.0, "high": 25.0}
    }
    
    # The memory is created here, not inside the class!
    # Inicializamos usando o dicionário e a config:
    pf_memory_particles = pf_estimator.initialize_particles(pf_config, world_prior_params)

    # --- Tracking Data for Plotting ---
    history = MultiUAVHistory()
    print("Starting Mission...")

    # --- NOVO: Preparando a Visualização Dinâmica ---
    # Calculamos o Z (Ground Truth) antes do loop para não recalcular a cada frame
    grid_x, grid_y = np.meshgrid(np.linspace(0, 50, 100), np.linspace(0, 50, 100))
    Z = gaussian_plume_gp((grid_x, grid_y, 2.5), source_position, 1.0, 5.0, (0.3, 0.3), wind_angles_gt)
    
    plt.ion() # Ativa o modo interativo do Matplotlib
    plt.figure(figsize=(10, 8))
    
    iteration = 0
    plot_interval = 20 # Atualiza o gráfico a cada 20 passos de simulação
    
    # --- 4. Mission Loop ---
    mission_active = True
    while mission_active:

        # STEP 1: Read measurements from all UAVs
        uav_data_snapshot = {}

        pos = uav1.get_position()
        real_conc = world.get_real_concentration(pos)

        sensor_data = {
            "MQ4": {
                "value": uav1.get_sensor("MQ4").read_measurement(real_conc),
                "noise_std": uav1.get_sensor("MQ4").noise_std
            },
            "CO2_Sensor": {
                "value": uav1.get_sensor("CO2_Sensor").read_measurement(real_conc),
                "noise_std": uav1.get_sensor("CO2_Sensor").noise_std
            }
        }

        uav_data_snapshot[uav1.id] = {
            "position": pos,
            "sensors": sensor_data
        }

        history.record_snapshot(uav_data_snapshot)

        # STEP 2: Execute Estimator
        pf_memory_particles = pf_estimator.step(
            history=history,
            memory=pf_memory_particles,
            config=pf_config
        )

        # --- NOVO: STEP 2.5: Real-time Visualization ---
        if iteration % plot_interval == 0:
            traj = history.get_trajectory(uav1.id)
            plot_particle_filter_step(
                pf_memory_particles, traj, 
                grid_x, grid_y, Z, source_position, 
                iteration
            )
        iteration += 1
        
        # STEP 3: Stop Criteria
        if t_progression >= 1.0:
            print("Mission Objective Reached: Path Complete.")
            mission_active = False
            break

        # STEP 4: Movement
        t_progression = min(t_progression + t_step, 1.0)
        next_pos = uav1.strategy.compute_next_step(t_progression)
        uav1.set_position(next_pos)

    # --- 5. Final Visualization ---
    grid_x, grid_y = np.meshgrid(np.linspace(0, 50, 200), np.linspace(0, 50, 200))
    Z = gaussian_plume_gp((grid_x, grid_y, 2.5), source_position, 1.0, 5.0, (0.3, 0.3), wind_angles_gt)

    # --- Extraindo os dados do UAV_01 para a plotagem legada ---
    traj_history = history.get_trajectory(uav1.id)
    meas_history = history.get_sensor_history(uav1.id, "MQ4")
    
    plot_mission_results(
        grid_x, grid_y, Z,
        traj_history, meas_history,
        source_position, wind_angles_gt
    )

    # --- Estimativa da taxa de emissão da fonte ---
    mean_q, std_q = plot_q_estimate(
        pf_memory_particles,
        true_q=env_config.q,
        prior_params=world_prior_params["q"]
    )
    print(f"q estimado: {mean_q:.4f} +/- {std_q:.4f} g/s")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
from q_study import weighted_median

runs = all_results["bem especificado"]["Q=10"]
rows = []
for r in runs:
    p = r["particles"]
    w = p["weight"] / p["weight"].sum()
    x = weighted_median(p["x_s"], p["weight"])
    y = weighted_median(p["y_s"], p["weight"])
    err = np.hypot(x - 10, y - 10)
    sd = np.sqrt(np.sum(w * ((p["x_s"] - x) ** 2 + (p["y_s"] - y) ** 2)))
    yaw = np.degrees(np.angle(np.sum(w * np.exp(1j * p["yaw"]))))
    rows.append((err, sd, yaw, weighted_median(p["zeta1"], p["weight"]), weighted_median(p["u_s"], p["weight"])))
rows = np.array(rows)

dif = (rows[:, 2] - 45 + 180) % 360 - 180
print(f"erro mediano: {np.median(rows[:,0]):.1f} m | desvio da posterior (mediano): {np.median(rows[:,1]):.1f} m")
print(f"seeds com yaw a menos de 30° da verdade (45°): {np.mean(np.abs(dif) < 30) * 100:.0f}%")
print(f"yaw em graus (p10, p50, p90): {np.round(np.percentile(rows[:,2], [10, 50, 90]))}")
print(f"ζ1 mediano: {np.median(rows[:,3]):.2f} (verdade 0,3) | u_s mediano: {np.median(rows[:,4]):.2f} (verdade 5)")
```

```python
import numpy as np
from experiments import run_experiments, DEFAULT_SCENARIO
from DispersionModels import gaussian_plume_gp
from q_study import PRIORS_FISICA_OK, PRIORS_Q_LARGA

P_ok = {**PRIORS_FISICA_OK, "q": PRIORS_Q_LARGA["q"]}
sc = dict(q=10.0, prior_params=P_ok, pf_source_height=1.5)
runs = run_experiments({"Q=10": sc}, seeds=range(50), keep_particles=True)["Q=10"]

full = {**DEFAULT_SCENARIO, **sc}
sigma = full["mq4_noise_std"]
h = full["pf_source_height"]          # altura da fonte nos dois lados (1,5 neste cenário)

def loglik_truth(r):
    t = np.array(r["trajectory"])
    pred = gaussian_plume_gp((t[:, 0], t[:, 1], t[:, 2]), (10.0, 10.0, h),
                             full["q"], full["u"], full["zetas"], full["wind_angles"])
    return -0.5 * np.sum(((r["mq4"] - pred) / sigma) ** 2)

def loglik_particles(r):
    t = np.array(r["trajectory"]); p = r["particles"]
    pred = gaussian_plume_gp((t[:, 0, None], t[:, 1, None], t[:, 2, None]),
                             (p["x_s"], p["y_s"], h), p["q"], p["u_s"],
                             (p["zeta1"], p["zeta2"]), (p["pitch"], p["yaw"]))
    return -0.5 * np.sum(((r["mq4"][:, None] - pred) / sigma) ** 2, axis=0)

d_top, d_best = [], []
for r in runs:
    ll, ll_t = loglik_particles(r), loglik_truth(r)
    d_top.append(ll[np.argmax(r["particles"]["weight"])] - ll_t)
    d_best.append(np.nanmax(ll) - ll_t)

print(f"partícula de maior peso vs verdade: mediana {np.median(d_top):+.1f} | seeds melhores que a verdade: {np.mean(np.array(d_top) > 0)*100:.0f}%")
print(f"melhor partícula do conjunto vs verdade: mediana {np.median(d_best):+.1f}")
```
